```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.feature_selection import SelectKBest, f_classif
```

```python
# Load dataset
df = pd.read_csv("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/outputs/balanced.csv")
```

```python
# Separate features and target
X = df.drop("Physical_Activity_Level", axis=1)
y = df["Physical_Activity_Level"]
```

```python
# Apply SelectKBest to select 7 best features
selector = SelectKBest(score_func=f_classif, k=7)
X_new = selector.fit_transform(X, y)
```

```python
# Get selected feature names
selected_features = X.columns[selector.get_support()]
print("Selected Features:", selected_features.tolist())
```

```python
# Plot feature importance scores
feature_scores = pd.DataFrame({
    "Feature": X.columns,
    "Score": selector.scores_
}).sort_values(by="Score", ascending=False)

sns.barplot(x="Score", y="Feature", data=feature_scores)
plt.title("Feature Importance Scores (SelectKBest)")
plt.savefig("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/eda_visualization/feature_importance_scores.png")
plt.clf()
```

```python
# Correlation heatmap of selected features
sns.heatmap(df[selected_features].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Selected Features")
plt.savefig("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/eda_visualization/correlation_selected_features.png")
plt.clf()
```

```python
# Create final dataset (7 features + target)
final_df = df[selected_features].copy()
final_df["Physical_Activity_Level"] = y.values
```

```python
# Save CSV
final_df.to_csv("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/outputs/step6_selected_features.csv", index=False)
```
