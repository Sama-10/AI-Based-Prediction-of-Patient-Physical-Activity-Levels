```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from imblearn.over_sampling import SMOTE
```

```python
# Load dataset
df = pd.read_csv("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/outputs/scaled.csv")
```

```python
# Split into features (X) and target (y)
X = df.drop("Physical_Activity_Level", axis=1)
y = df["Physical_Activity_Level"]
```

```python
#Plot class distribution before balancing
sns.countplot(x=y)
plt.title("Physical Activity Level - Before SMOTE")
plt.savefig("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/eda_visualization/activity_level_before_smote.png")
plt.clf()
```

```python
#Apply SMOTE
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X, y)
```

```python
#Plot class distribution after balancing
sns.countplot(x=y_res)
plt.title("Physical Activity Level - After SMOTE")
plt.savefig("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/outputs/activity_level_after_smote.png")
plt.clf()
```
