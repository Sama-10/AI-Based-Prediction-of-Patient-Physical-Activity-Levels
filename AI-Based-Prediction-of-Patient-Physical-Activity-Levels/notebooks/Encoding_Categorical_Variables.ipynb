```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
```

```python
# Load dataset 
df = pd.read_csv("C:\\Users\\USER\\Desktop\\2025-Y2-S1-MLB-B5G1-04\\Data\\Health.csv")
```

```python
# Visualize categorical distributions before encoding
sns.countplot(x="Smoking_Status", data=df)
plt.title("Smoking Status Distribution (Before Encoding)")
plt.savefig("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/eda_visualization/smoking_status_before.png")
plt.clf()

sns.countplot(x="Physical_Activity_Level", data=df)
plt.title("Physical Activity Level Distribution (Before Encoding)")
plt.savefig("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/eda_visualization/activity_level_before.png")
plt.clf()
```

```python
# Encode Smoking_Status → binary mapping
df["Smoking_Status"] = df["Smoking_Status"].map({"non-smoker": 0, "smoker": 1})
```

```python
df.head(5)
```

```python
# Encode Physical_Activity_Level → label encoding
le = LabelEncoder()
df["Physical_Activity_Level"] = le.fit_transform(df["Physical_Activity_Level"])
```

```python
df.head(5)
```

```python
#Visualize categorical distributions after encoding
sns.countplot(x="Smoking_Status", data=df)
plt.title("Smoking Status Distribution (After Encoding)")
plt.savefig("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/eda_visualization/smoking_status_after.png")
plt.clf()

sns.countplot(x="Physical_Activity_Level", data=df)
plt.title("Physical Activity Level Distribution (After Encoding)")
plt.savefig("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/eda_visualization/activity_level_after.png")
plt.clf()
```

```python
# Save encoded dataset
df.to_csv("C:/Users/USER/Desktop/2025-Y2-S1-MLB-B5G1-04/Results/outputs/encoded.csv", index=False)
```
